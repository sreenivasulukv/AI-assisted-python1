Example for Break statement

```python
List_of_books=["python","Java", "C++"]
book=input("enter book you want in library")
book_no= 0 # start from index "Zero"
while(book_no < len(List_of_books)):   # Loop condition corrected to avoid out of bounds access
  if(book==List_of_books[book_no]):
    print("yes, book available")
    break
  book_no=book_no+1
else:
  print("No, book not available")
```

### print number table using for

```python
num = int(input("Enter the number for the table you want to print: "))
for i in range(1, 11):
  print(f"{num} x {i} = {num * i}")
```

```python
print("Demonstrating 'for' loop with 'break' statement:")
for i in range(10):
  if i == 5:
    print(f"Breaking the loop when i is {i}")
    break
  print(f"Current number: {i}")
print("Loop finished.")
```
