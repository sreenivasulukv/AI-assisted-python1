## Python Data Types

In Python, data types are used to classify data and determine the operations that can be performed on them. Here are some common built-in data types and how to print their types.

```python
# Integer
num_int = 10
print(f"Value: {num_int}, Type: {type(num_int)}\n")

# Float
num_float = 10.5
print(f"Value: {num_float}, Type: {type(num_float)}\n")

# String
text_str = "Hello, Python!"
print(f"Value: '{text_str}', Type: {type(text_str)}\n")

# List
list_data = [1, 2, 3]
print(f"Value: {list_data}, Type: {type(list_data)}\n")

# Tuple
tuple_data = (1, 2, 3)
print(f"Value: {tuple_data}, Type: {type(tuple_data)}\n")

# Dictionary
dict_data = {'a': 1, 'b': 2}
print(f"Value: {dict_data}, Type: {type(dict_data)}\n")

# Boolean
bool_data = True
print(f"Value: {bool_data}, Type: {type(bool_data)}\n")
```
