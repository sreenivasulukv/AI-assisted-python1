## Python Data Types

In Python, data types are used to classify data and determine the operations that can be performed on them. Here are some common built-in data types and how to print their types.

```python
# Integer
num_int = 10
print(f"Value: {num_int}, Type: {type(num_int)}\n")

# Float
num_float = 10.5
print(f"Value: {num_float}, Type: {type(num_float)}\n")

# String
text_str = "Hello, Python!"
print(f"Value: '{text_str}', Type: {type(text_str)}\n")

# List
list_data = [1, 2, 3]
print(f"Value: {list_data}, Type: {type(list_data)}\n")

# Tuple
tuple_data = (1, 2, 3)
print(f"Value: {tuple_data}, Type: {type(tuple_data)}\n")

# Dictionary
dict_data = {'a': 1, 'b': 2}
print(f"Value: {dict_data}, Type: {type(dict_data)}\n")

# Boolean
bool_data = True
print(f"Value: {bool_data}, Type: {type(bool_data)}\n")
```

## Simple Calculator Program (Line by Line)

```python
# 1. Get the first number from the user.
# input() reads a string, int() converts it to an integer.
num1 = int(input("Enter first number: "))
```

```python
# 2. Get the second number from the user.
# Similar to the first number, convert it to an integer.
num2 = int(input("Enter second number: "))
```

```python
# 3. Get the operation choice from the user.
# The input is expected to be a string like '+', '-', '*', or '/'.
operation = input("Enter operation (+, -, *, /): ")
```

```python
# 4. Use if-elif-else statements to perform the chosen operation.
# Check the 'operation' variable and execute the corresponding code block.

# If the user entered '+', perform addition.
if operation == '+':
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")

# Else if the user entered '-', perform subtraction.
elif operation == '-':
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")

# Else if the user entered '*', perform multiplication.
elif operation == '*':
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")

# Else if the user entered '/', perform division.
elif operation == '/':
    # Check if the second number is not zero to avoid division by zero error.
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Error: Cannot divide by zero.")

# If none of the above operations match, it's an invalid input.
else:
    print("Invalid operation. Please choose from +, -, *, or /.")
```

## Calculator Program in a Single Code Cell

```python
# Get the first number from the user.
num1 = int(input("Enter first number: "))

# Get the second number from the user.
num2 = int(input("Enter second number: "))

# Get the operation choice from the user.
operation = input("Enter operation (+, -, *, /): ")

# Use if-elif-else statements to perform the chosen operation.
if operation == '+':
    result = num1 + num2
    print(f"{num1} + {num2} = {result}")

elif operation == '-':
    result = num1 - num2
    print(f"{num1} - {num2} = {result}")

elif operation == '*':
    result = num1 * num2
    print(f"{num1} * {num2} = {result}")

elif operation == '/':
    if num2 != 0:
        result = num1 / num2
        print(f"{num1} / {num2} = {result}")
    else:
        print("Error: Cannot divide by zero.")

else:
    print("Invalid operation. Please choose from +, -, *, or /.")
```
